In c,adding a new field to a struct requires changing the struct defination.what happens in python when you add a new key to a dictionary while the program is running-and why is that both useful and risky?
Ans:Why is this useful?
Flexible: You can add information whenever you need it.
No fixed structure: Different dictionaries can have different keys.
Useful when working with dynamic data, such as JSON or API responses.
Why is this risky?
There is no requirement that every student dictionary has the same fields.
Missing keys can cause errors when you try to access them.

In short: Python dictionaries are more flexible than C structs, but that flexibility can make the data less predictable and easier to misuse.

The c program that marks is a float.The python dictionary does not declare anything.where does the type come from?
Ans:In Python, the type comes from the value you assign to the dictionary key.

For example:

student["marks"] = 87.5

Because 87.5 is a float, Python treats student["marks"] as a float.

Similarly:

student["roll"] = 101      # int
student["name"] = "Rahul"  # str

So Python uses dynamic typing: the variable/key doesn't have a fixed declared type; the object/value itself has a type.

C: type is declared → float marks
Python: type is inferred at runtime → 87.5 → float
